## A missão
O trio escolhe um assunto, **busca dados reais na internet**, encontra **um achado** e conta essa história para a turma, **neste notebook**, em **3 minutos**.

Um número solto não diz nada. Uma história com dados tem três coisas: uma **pergunta**, um **achado em destaque** e um **"e daí?"** (o que o público deve pensar ou fazer).

| Horário | O que acontece |
|---|---|
| 19h00 | Ler as instruções e escolher o assunto |
| 19h20 | Buscar os dados e montar a história (Passos 1 a 6) |
| 21h00 | Apresentações: 3 minutos por trio |

**Como trabalhar:** este notebook só tem instruções. Abaixo de cada passo, o trio cria as próprias células: **+ Código** para os comandos e **+ Texto** para as respostas.

## O roteiro: as 6 fases do CRISP-DM viram a história

| Passo | Fase do CRISP-DM | O que o trio decide |
|---|---|---|
| 1 | Entendimento do negócio | Qual é a pergunta? Para quem contamos? |
| 2 | Entendimento dos dados | Onde estão os dados? Qual é a fonte? |
| 3 | Preparação dos dados | Que linhas e colunas interessam? |
| 4 | Modelagem (aqui: contar e comparar) | Qual é o achado? |
| 5 | Avaliação | O achado está certo? O que ele **não** mostra? |
| 6 | Implantação | O gráfico final e a história falada |

## Um exemplo do que entregar

**Pergunta:** quantos brasileiros usam a internet, e quantos ainda estão fora? **Público:** uma secretaria que planeja inclusão digital. **Fonte:** Our World in Data (ourworldindata.org/grapher/share-of-individuals-using-the-internet).

**Gráfico:** uma linha de 2000 a 2024, com o Brasil em cor e o mundo em cinza. Título: *"Em 24 anos, o Brasil foi de 3% para 84% de pessoas conectadas"*.

**A história, em três falas (uma por integrante):**

- **Começo** (o contexto): *"Uma secretaria quer planejar a inclusão digital. Fomos ver quantos brasileiros usam a internet, com dados do Our World in Data."*
- **Meio** (o achado): *"Em 2000, só 3 em cada 100 brasileiros usavam a internet. Em 2024, são 84. O Brasil passou a média do mundo, mas ainda está atrás de Chile, Uruguai e Argentina."*
- **Fim** (o "e daí?"): *"Faltam 16 em cada 100 brasileiros. O próximo passo é descobrir quem são e onde estão, porque é para eles que a política precisa olhar."*

# Agora é com o trio

## Passo 1 · A pergunta e o público
Escolham um assunto que interesse ao trio: esporte, música, jogos, preços, clima, saúde, a cidade, a universidade. Transformem o assunto em **uma pergunta que uma tabela responde**.

Criem uma célula de texto com:

- os **integrantes** do trio;
- a **pergunta**;
- **para quem** contamos (quem se interessa por essa resposta?).

## Passo 2 · Conseguir os dados
Três jeitos simples. Usem **um**.

**Regra:** todo dado precisa de **fonte com link**. A IA **não é fonte de dados**: ela pode inventar números.

**Jeito A · Digitar uma tabela pequena.** Serve para histórias com poucos números (um ranking, 5 a 10 linhas). Achem a tabela numa página confiável (Wikipédia, IBGE, site oficial, reportagem que cita a fonte) e digitem os valores numa tabela do pandas.

**Jeito B · Ler um arquivo por link.** No **Our World in Data** (ourworldindata.org) há milhares de gráficos sobre saúde, energia, educação, internet, clima e população. Abram um gráfico, copiem o endereço da página, acrescentem **.csv** no final e leiam esse endereço com o pandas, como fizemos com os arquivos das outras práticas.

**Jeito C · Baixar um arquivo e enviar para o Colab.** Baixem um arquivo CSV ou Excel de um portal de dados. No Colab, cliquem no ícone de **pasta** 📁 (à esquerda), arrastem o arquivo para lá e leiam pelo nome. Onde procurar: **IBGE Cidades** (cidades.ibge.gov.br) · **dados.gov.br** · **Kaggle** (kaggle.com/datasets) · **Google Dataset Search** (datasetsearch.research.google.com) · portais de transparência.

Ao terminar, anotem numa célula de texto a **fonte** (nome do site e link) e o **ano** dos dados.

## Passo 3 · Preparar
Fiquem só com as linhas e colunas que respondem à pergunta. Quase sempre bastam três operações que vocês já usaram: **escolher colunas**, **filtrar linhas** e **resumir por grupo** (média, soma ou contagem).

Travou? Usem o **Gemini** do Colab (botão ✨): descrevam a tabela e o que querem, rodem o código sugerido e **confiram o resultado**.

## Passo 4 · O achado
Olhem a tabela pronta e procurem o que **foge do esperado**: o maior, o menor, a virada, a diferença grande.

**A IA como copiloto.** Se o trio estiver sem ideia, copiem a tabela (ou um print do gráfico) para o Gemini ou outro assistente, com este pedido:

> *O objetivo da nossa apresentação é: (escrevam a pergunta do Passo 1). Identifique as três informações mais importantes destes dados. Para cada uma, explique por que ela é relevante. Depois, faça perguntas que nos ajudem a transformar cada informação em um destaque: o motivo por trás dela e a ação que ela sugere.*

A IA ajuda a **pensar**; quem decide e confere é o trio. Ela erra contas e datas: **todo número da história tem de aparecer na saída de uma célula deste notebook**.

Escrevam numa célula de texto **o achado do trio, em uma frase com número**.

## Passo 5 · Conferir
Antes de desenhar, respondam **sim** às quatro perguntas:

1. O número do achado saiu de uma célula deste notebook, e não da cabeça ou da IA?
2. A fonte e o ano dos dados estão anotados?
3. Sabemos o que ficou **de fora** (linhas removidas, anos sem dado, grupos pequenos)?
4. A frase do achado **não** afirma causa? Os dados mostram que algo aconteceu, não por quê.

## Passo 6 · O gráfico e a história
Façam **um** gráfico. Para comparar categorias, **barras**; para mostrar mudança no tempo, **linha**. Ele precisa de:

- **um destaque só**: o ponto do achado em cor, o resto em cinza;
- **título que diz a conclusão**, não o assunto ("O Brasil foi de 3% para 84%", e não "Uso da internet");
- **barras começando no zero**;
- **fonte** no rodapé;
- nada que não ajude a mensagem (grade, legenda sobrando, cores sem significado).

Depois, escrevam numa célula de texto **a história do trio**, com duas ou três frases por fala e o nome de quem fala cada parte:

- **Começo:** o contexto (a pergunta, para quem, de onde vêm os dados);
- **Meio:** o achado, com os números;
- **Fim:** o "e daí?" (o que o público deve pensar ou fazer, e o que os dados não mostram).

## Apresentação
- **3 minutos** por trio, com **este notebook** na tela.
- Mostrem só o **gráfico final**; a história é **falada**, não lida.
- Os três falam: começo, meio e fim.

**Quem assiste observa:** entendi a mensagem em 5 segundos? O título diz a conclusão? Há um destaque só? A fonte aparece? A história termina com um "e daí?"
